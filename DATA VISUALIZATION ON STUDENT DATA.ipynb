# BAR CHART 

```python
gender_grade = df.groupby("sex")["G3"].mean()                       # AVG FINAL GRADE FOR EACH GENDER

plt.figure(figsize=(8, 5))                                          # CREATE BAR CHART

ax = gender_grade.plot(kind="bar")

plt.title("Average Final Grade by Gender")                          # TITLE

plt.xlabel("Gender")                                                # X- AXIS

plt.ylabel("Average Final Grade")                                   # Y -AXIS

plt.xticks(rotation=0)

for i, value in enumerate(gender_grade):                           # ADD NUM VALUE ABOVE EACH BAR 
    ax.text(i, value + 0.1, f"{value:.2f}", ha="center")
```

# LINE CHART 

```python
study_grade = df.groupby("studytime")["G3"].mean()                         # AVG FINAL GRADE FOR EACH GENDER

study_grade
```

```python
plt.figure(figsize=(8, 5))                                                        # CREATE THE FIG

plt.plot(study_grade.index, study_grade.values, marker="o")                       # CREATE THE LINE CHART

plt.title("Average Final Grade by Study Time")                                    # TITLE

plt.xlabel("Study Time Category")                                                 # X - AXIS
                                                                                
plt.ylabel("Average Final Grade")                                                 # Y - AXIS

plt.grid(True)                                                                    # ADD GRID LINES 

```

# SCATTER PLOT 

```python
# Relationship Between Student Absence and Final Grade 
```

```python
plt.figure(figsize=(8, 5))                                        # CREATE FIG

sns.scatterplot(data=df, x="absences", y="G3")                    # CREATE A SCATTERPLOT 

plt.title("Absences vs Final Grade")                              # TITLE

plt.xlabel("Number of Absences")                                  # X - LABEL

plt.ylabel("Final Grade")                                         # Y - LABEL 
```

# ADD A TREND LINE IN SCATTER PLOT

```python
plt.figure(figsize=(8, 5))                                                          # CREATE THE FIG

sns.regplot(data=df, x="absences", y="G3")                                          # CREATE SCATTER PLOT WITH TREND LINE 

plt.title("Relationship Between Absences and Final Grade")                          # TITLE 

plt.xlabel("Number of Absences")                                                    # X - AXIS 

plt.ylabel("Final Grade")                                                           # Y - AXIS
```

# HEATMAP 

```python
numeric_data = df[
    ["G3", "studytime", "absences", "Medu"]                         # SELECT IMPORTANT NUM COLUMN
]

correlation = numeric_data.corr()                                   # CALCULATE CORRELATION BTW VARIABLES 

correlation
```

```python
plt.figure(figsize=(8, 6))                                                             # CREATE FIG

sns.heatmap(                                                                           # CREATE HEATMAP
    correlation,
    annot=True,
    cmap="YlGnBu",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Student Performance Variables")                      # TITLE
```

# BOXPLOT           STUDY TIME VS FINAL GRADE 

```python
plt.figure(figsize=(8, 5))                                              # CREATE FIG

sns.boxplot(data=df, x="studytime", y="G3")                             # CREATE BOXPLOT FOR EACH STUDY TIME CATEGORY 

plt.title("Distribution of Final Grades by Study Time")                 # TITLE

plt.xlabel("Study Time Category")                                       # X - AXIS 

plt.ylabel("Final Grade")                                               # Y - AXIS
```
